```python
import numpy as np
import matplotlib.pyplot as plt

from .model import MLP


np.random.seed(1337)
```

```python
# Flattened input image, where image has dims 28 x 28 
x = np.random.randn(784, 1)

# Say correct output class is 3
y = np.zeros((10, 1))
y[3] = 1
```

```python
plt.imshow(x.reshape(28, 28)) 
plt.title("Input Image")
plt.yticks([])
plt.xticks([])
plt.show()



plt.title("Correct Class")
plt.imshow(y)
plt.yticks(np.arange(0, 10))
plt.xticks([])
plt.show()

```

```python
mlp = MLP()

initial_prediction = mlp.forward(x)

plt.title("Initial Output Prediction")
plt.imshow(initial_prediction)
plt.yticks(np.arange(0, 10))
plt.xticks([])
plt.show()
```

```python
# Train the model 1000 times on the same vector and visualise the final prediction
for i in range(1000):
    y_pred = mlp.forward(x)
    grads = mlp.back(y)
    mlp.step(grads, 1e-3)




plt.title("Final Class Prediction after 1000 training loops")
plt.imshow(y_pred)
plt.yticks(np.arange(0, 10))
plt.xticks([])
plt.show()
```
